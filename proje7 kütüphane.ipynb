```python
import sqlite3
import pandas as pd
import numpy as np 
from IPython.display import display 



conn = sqlite3.connect('kutuphane.db')
cursor = conn.cursor()


cursor.execute('''
CREATE TABLE IF NOT EXISTS kitaplar (
    id INTEGER PRIMARY KEY,
    baslik TEXT NOT NULL,
    yazar TEXT NOT NULL,
    yayinevi TEXT
);
''')


cursor.execute("INSERT OR IGNORE INTO kitaplar (id, baslik, yazar, yayinevi) VALUES (1, 'Sefiller', 'Victor Hugo', 'Can Yayınları')")
cursor.execute("INSERT OR IGNORE INTO kitaplar (id, baslik, yazar, yayinevi) VALUES (2, 'Suç ve Ceza', 'Fyodor Dostoyevski', 'İletişim Yayınları')")
cursor.execute("INSERT OR IGNORE INTO kitaplar (id, baslik, yazar, yayinevi) VALUES (3, 'Küçük Prens', 'Antoine de Saint-Exupéry', 'Mavi Bulut Yayınları')")
cursor.execute("INSERT OR IGNORE INTO kitaplar (id, baslik, yazar, yayinevi) VALUES (4, 'Don Kişot', 'Miguel de Cervantes', 'Yapı Kredi Yayınları')")

conn.commit()



df_kitaplar = pd.read_sql_query("SELECT * FROM kitaplar", conn)
conn.close()



df_kitaplar['sayfa_sayisi'] = np.random.randint(100, 500, size=len(df_kitaplar))



df_kitaplar['ulke'] = np.random.choice(['Türkiye', 'İngiltere', 'Fransa'], size=len(df_kitaplar))


print("--- Geçici Sütunlar Eklendi (Test Verisi) ---")
display(df_kitaplar.head())



df_filtreli = df_kitaplar[(df_kitaplar['sayfa_sayisi'] > 250) & (df_kitaplar['ulke'] == 'Türkiye')]


print("\n--- Sayfa Sayısı > 250 ve Ülkesi Türkiye Olan Kitaplar ---")
display(df_filtreli)
```
